```python
import os, sys
sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), '..')))
```

```python
from pathlib import Path
from tqdm import tqdm
from PIL import Image
import numpy as np

from swmf.data_gen_corrupt import (
    MAX_SEVERITY,
    gaussian_noise,
    shot_noise,
    impulse_noise,
    speckle_noise,
    gaussian_blur,
    defocus_blur,
    frosted_blur,
    zoom_blur,
    fog,
    brightness,
    contrast,
    saturate,
    pixelate,
    jpeg_compression,
)
```

```python
LARD_TEST_PATH = Path("../data/datasets/lard_512x512_ICPR2026/images/test").resolve()
LARD_TEST_PATH
```

```python
MAX_N_IMG = -1
```

Create the dataset of corrupted data `test_corrupt`

```python
### Reproducibility ###
SEED = 42
np.random.seed(SEED)
#######################
```

```python
src_path = LARD_TEST_PATH
dst_path = src_path.parent / (src_path.stem + "_corrupt")
dst_path.mkdir(exist_ok=True, parents=True)

image_files = [f for f in os.listdir(src_path) if f.endswith(('.png', '.jpg', '.jpeg'))]
image_files.sort()
image_files = image_files[:MAX_N_IMG]
corruptions = {
    'Gaussian_Noise': gaussian_noise,
    # 'Shot_Noise': shot_noise,
    # 'Impulse_Noise': impulse_noise,
    # 'Speckle_Noise': speckle_noise,
    # 'Gaussian_Blur': gaussian_blur,
    # 'Defocus_Blur': defocus_blur,
    'Frosted_Blur': frosted_blur,
    # 'Zoom_Blur': zoom_blur,
    'Fog': fog,
    # 'Brightness': brightness,
    # 'Contrast': contrast,
    # 'Saturate': saturate,
    # 'Pixelate': pixelate,
    # 'JPEG': jpeg_compression,
}

for corrupt_id, corrupt_fn in corruptions.items():
    print(f"Creating images for corruption: {corrupt_id}")

    for severity in range(1, MAX_SEVERITY + 1):
        print(f"Severity {severity}")

        severity_dst_path = dst_path / f"{corrupt_id}_severity_{severity}"
        severity_dst_path.mkdir(exist_ok=True, parents=True)

        for img_file in tqdm(image_files):
            img_path = src_path / img_file
            img = Image.open(img_path)
            corrupted_img = corrupt_fn(np.array(img), severity)
            corrupted_img = Image.fromarray(corrupted_img.astype('uint8'))
            corrupted_img.save(severity_dst_path / img_file)
```
